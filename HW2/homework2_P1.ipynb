## A simple agent using tools from the MCP server


```python
import re
import json
from difflib import SequenceMatcher
from typing import Any, Dict, List, Tuple, Optional

from langchain_core.messages import HumanMessage, SystemMessage
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_mcp_adapters.client import MultiServerMCPClient

client = MultiServerMCPClient({
    "social_graph": {
        "transport": "http",
        "url": "https://ftec5660.ngrok.app/mcp",
        "headers": {"ngrok-skip-browser-warning": "true"}
    }
})
mcp_tools = await client.get_tools()

def _norm_tool_name(n: str) -> str:
    return re.sub(r"[^a-z0-9]+", "_", (n or "").strip().lower()).strip("_")

def _build_tool_map(tools) -> Dict[str, Any]:
    tool_map = {}
    for t in tools:
        nn = _norm_tool_name(getattr(t, "name", ""))

        if "facebook" in nn and "search" in nn:
            tool_map["fb_search"] = t
        elif "facebook" in nn and "profile" in nn and "mutual" not in nn:
            tool_map["fb_profile"] = t
        elif "facebook" in nn and "mutual" in nn:
            tool_map["fb_mutual"] = t

        elif "linkedin" in nn and "search" in nn:
            tool_map["li_search"] = t
        elif "linkedin" in nn and "profile" in nn and "interaction" not in nn:
            tool_map["li_profile"] = t
        elif "linkedin" in nn and "interaction" in nn:
            tool_map["li_interactions"] = t
    return tool_map

TOOL_MAP = _build_tool_map(mcp_tools)
print("Detected tool map:", {k: getattr(v, "name", None) for k, v in TOOL_MAP.items()})

assert "li_search" in TOOL_MAP and "li_profile" in TOOL_MAP
assert "fb_search" in TOOL_MAP and "fb_profile" in TOOL_MAP


# 1) LLM init
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=GEMINI_VERTEX_API_KEY,
    temperature=0,
)


# 2) Robust unwrap for MCP tool outputs
def unwrap_mcp(obj: Any) -> Any:
    """
    Handles these cases:
    A) dict wrapper: {"type":"text","text":"<json>", "id":"lc_..."}
    B) list wrapper: [ {wrapper}, {wrapper}, ... ]
    C) plain JSON string
    Returns a Python dict/list/primitive.
    """
    # Case B: list -> unwrap each element; if it becomes a single-element wrapper, unwrap again
    if isinstance(obj, list):
        unwrapped = [unwrap_mcp(x) for x in obj]
        # if tool returns [<actual_list>] or [<actual_dict>], flatten one layer
        if len(unwrapped) == 1 and isinstance(unwrapped[0], (list, dict)):
            return unwrapped[0]
        return unwrapped

    # Case A: dict wrapper
    if isinstance(obj, dict) and obj.get("type") == "text" and "text" in obj:
        txt = obj["text"]
        # text is JSON string, parse it
        parsed = json.loads(txt)
        # parsed may itself be list of wrapper dicts (rare), unwrap again
        return unwrap_mcp(parsed)

    # Case C: JSON string directly
    if isinstance(obj, str):
        s = obj.strip()
        if (s.startswith("{") and s.endswith("}")) or (s.startswith("[") and s.endswith("]")):
            parsed = json.loads(s)
            return unwrap_mcp(parsed)
        return obj

    return obj


# 3) CV extraction
EXTRACT_SYS = """You extract structured candidate info from a CV text.
Return ONLY valid JSON, no markdown, no extra text.

Schema:
{
  "name": str|null,
  "location": {"city": str|null, "country": str|null},
  "education": [{"school": str|null, "degree": str|null, "field": str|null, "start_year": int|null, "end_year": int|null}],
  "experience": [{"company": str|null, "title": str|null, "start_year": int|null, "end_year": int|null}],
  "skills": [str, ...],
  "keywords": [str, ...]
}

Rules:
- If unknown, use null.
- years must be integers or null.
- skills <= 30, keywords <= 25 (dedup).
"""

def _safe_json_load(s: str) -> Dict[str, Any]:
    m = re.search(r"\{.*\}", s, flags=re.S)
    if m:
        return json.loads(m.group(0))
    return json.loads(s)

def _dedup_list(xs: List[Any], cap: int) -> List[str]:
    seen, out = set(), []
    for x in xs or []:
        x2 = str(x).strip()
        if not x2:
            continue
        k = x2.lower()
        if k in seen:
            continue
        seen.add(k)
        out.append(x2)
        if len(out) >= cap:
            break
    return out

async def extract_cv_struct(cv_text: str) -> Dict[str, Any]:
    msg = [SystemMessage(content=EXTRACT_SYS), HumanMessage(content=cv_text[:20000])]
    out = llm.invoke(msg)
    raw = out.content if hasattr(out, "content") else str(out)
    data = _safe_json_load(raw)

    if "location" not in data or data["location"] is None:
        data["location"] = {"city": None, "country": None}
    data.setdefault("education", [])
    data.setdefault("experience", [])
    data.setdefault("skills", [])
    data.setdefault("keywords", [])

    data["skills"] = _dedup_list(data["skills"], 30)
    data["keywords"] = _dedup_list(data["keywords"], 25)
    return data

# 4) Similarity + scoring
def sim(a: Optional[str], b: Optional[str]) -> float:
    if not a or not b:
        return 0.0
    return SequenceMatcher(None, str(a).strip().lower(), str(b).strip().lower()).ratio()

def normalize_loc(city: Optional[str], country: Optional[str]) -> str:
    parts = []
    if city: parts.append(str(city).strip().lower())
    if country: parts.append(str(country).strip().lower())
    return ", ".join(parts)

def year_overlap(a_start, a_end, b_start, b_end) -> bool:
    if a_start is None and a_end is None: return False
    if b_start is None and b_end is None: return False
    a_s = a_start if a_start is not None else -10**9
    a_e = a_end if a_end is not None else 10**9
    b_s = b_start if b_start is not None else -10**9
    b_e = b_end if b_end is not None else 10**9
    return not (a_e < b_s or b_e < a_s)

def _skill_set(skills: List[str]) -> set:
    return set(re.sub(r"\s+", " ", s.strip().lower()) for s in (skills or []) if s and str(s).strip())

def _skills_score(cv_skills: List[str], li_skills: List[Dict[str, Any]]) -> Tuple[float, List[str]]:
    flags = []
    cv_set = _skill_set(cv_skills)
    li_set = _skill_set([s.get("name") for s in (li_skills or []) if isinstance(s, dict)])

    if not cv_set and not li_set:
        return 0.1, ["No skills found on both CV and profile"]
    if not li_set:
        return 0.0, ["Profile has no skills section"]
    if not cv_set:
        return 0.15, ["CV has no skills section"]

    inter = len(cv_set & li_set)
    union = len(cv_set | li_set)
    jacc = inter / union if union else 0.0
    if jacc < 0.12:
        flags.append("Skills overlap is low")
    return jacc, flags

def _edu_match_score(cv_edu: List[Dict[str, Any]], li_edu: List[Dict[str, Any]]) -> Tuple[float, List[str]]:
    flags = []
    if not cv_edu:
        return 0.2, flags
    if not li_edu:
        return 0.0, ["Profile missing education section"]

    best = 0.0
    for ce in cv_edu:
        local_best = 0.0
        for le in li_edu:
            s_school = sim(ce.get("school"), le.get("school"))
            s_degree = sim(ce.get("degree"), le.get("degree"))
            s_field  = sim(ce.get("field"),  le.get("field"))
            s_year   = 1.0 if year_overlap(ce.get("start_year"), ce.get("end_year"), le.get("start_year"), le.get("end_year")) else 0.0
            score = 0.55*s_school + 0.15*s_degree + 0.15*s_field + 0.15*s_year
            if score > local_best:
                local_best = score
        if local_best > best:
            best = local_best

    if best < 0.35:
        flags.append("Education mismatch (low similarity)")
    return best, flags

def _exp_match_score(cv_exp: List[Dict[str, Any]], li_exp: List[Dict[str, Any]]) -> Tuple[float, List[str]]:
    flags = []
    if not cv_exp:
        return 0.15, ["CV missing experience section"]
    if not li_exp:
        return 0.0, ["Profile missing experience section"]

    matched = 0
    total = min(len(cv_exp), 6)
    for ce in cv_exp[:total]:
        best = 0.0
        for le in li_exp:
            s_company = sim(ce.get("company"), le.get("company"))
            s_title   = sim(ce.get("title"),   le.get("title"))
            s_year    = 1.0 if year_overlap(ce.get("start_year"), ce.get("end_year"), le.get("start_year"), le.get("end_year")) else 0.0
            score = 0.55*s_company + 0.25*s_title + 0.20*s_year
            if score > best:
                best = score
        if best >= 0.45:
            matched += 1

    ratio = matched / max(total, 1)
    if ratio < 0.34:
        flags.append(f"Experience weak match ({matched}/{total} lines matched)")
    return ratio, flags

def _name_loc_score(cv: Dict[str, Any], prof: Dict[str, Any], name_key: str, city_key: str, country_key: str) -> Tuple[float, List[str]]:
    flags = []
    s_name = sim(cv.get("name"), prof.get(name_key))

    cv_loc = normalize_loc(cv.get("location", {}).get("city"), cv.get("location", {}).get("country"))
    pr_loc = normalize_loc(prof.get(city_key), prof.get(country_key))
    s_loc = sim(cv_loc, pr_loc) if (cv_loc and pr_loc) else 0.15

    if s_name < 0.55:
        flags.append("Name mismatch")
    if cv_loc and pr_loc and s_loc < 0.45:
        flags.append("Location mismatch")

    return 0.75*s_name + 0.25*s_loc, flags


# 5) Tool calls
async def search_linkedin_candidates(cv: Dict[str, Any], limit: int = 8) -> List[Dict[str, Any]]:
    tool = TOOL_MAP["li_search"]
    name = (cv.get("name") or "").strip()
    res = await tool.ainvoke({"q": name, "limit": limit, "fuzzy": True})
    res = unwrap_mcp(res)
    assert isinstance(res, list), f"LinkedIn search unexpected type: {type(res)}"
    return res[:limit]

async def get_linkedin_profile(person_id: int) -> Dict[str, Any]:
    tool = TOOL_MAP["li_profile"]
    prof = await tool.ainvoke({"person_id": int(person_id)})
    prof = unwrap_mcp(prof)
    assert isinstance(prof, dict), f"LinkedIn profile unexpected type: {type(prof)}"
    return prof

async def get_linkedin_interactions(person_id: int) -> Dict[str, Any]:
    tool = TOOL_MAP["li_interactions"]
    inter = await tool.ainvoke({"person_id": int(person_id)})
    inter = unwrap_mcp(inter)
    assert isinstance(inter, dict), f"LinkedIn interactions unexpected type: {type(inter)}"
    return inter

async def search_facebook_candidates(cv: Dict[str, Any], limit: int = 8) -> List[Dict[str, Any]]:
    tool = TOOL_MAP["fb_search"]
    name = (cv.get("name") or "").strip()
    res = await tool.ainvoke({"q": name, "limit": limit, "fuzzy": True})
    res = unwrap_mcp(res)
    assert isinstance(res, list), f"Facebook search unexpected type: {type(res)}"
    return res[:limit]

async def get_facebook_profile(user_id: int) -> Dict[str, Any]:
    tool = TOOL_MAP["fb_profile"]
    prof = await tool.ainvoke({"user_id": int(user_id)})
    prof = unwrap_mcp(prof)
    assert isinstance(prof, dict), f"Facebook profile unexpected type: {type(prof)}"
    return prof


# 6) Scoring
def score_linkedin(cv: Dict[str, Any], li_profile: Dict[str, Any], li_inter: Dict[str, Any]) -> Tuple[float, List[str]]:
    flags = []

    s_name_loc, f1 = _name_loc_score(cv, li_profile, "name", "city", "country")
    flags += f1

    s_edu, f2 = _edu_match_score(cv.get("education", []), li_profile.get("education", []))
    flags += f2

    s_exp, f3 = _exp_match_score(cv.get("experience", []), li_profile.get("experience", []))
    flags += f3

    s_skill, f4 = _skills_score(cv.get("skills", []), li_profile.get("skills", []))
    flags += f4

    post_count = li_inter.get("post_count", 0) or 0
    engagement = li_inter.get("engagement_score", 0.0) or 0.0
    bonus = 0.0
    if post_count == 0 and engagement == 0:
        flags.append("LinkedIn engagement very low (weak evidence)")
        bonus -= 0.03
    elif engagement >= 2.0:
        bonus += 0.03

    base = 0.30*s_name_loc + 0.25*s_exp + 0.25*s_edu + 0.20*s_skill
    score = max(0.0, min(1.0, base + bonus))

    hard_penalty = 0.0
    if s_name_loc < 0.35: hard_penalty += 0.25
    if s_exp < 0.20:      hard_penalty += 0.15
    if s_edu < 0.20 and cv.get("education"): hard_penalty += 0.10

    score = max(0.0, min(1.0, score - hard_penalty))
    return float(score), flags

def score_facebook(cv: Dict[str, Any], fb_profile: Dict[str, Any]) -> Tuple[float, List[str]]:
    flags = []
    s_name = max(sim(cv.get("name"), fb_profile.get("display_name")),
                 sim(cv.get("name"), fb_profile.get("original_name")))

    cv_loc = normalize_loc(cv.get("location", {}).get("city"), cv.get("location", {}).get("country"))
    fb_loc = normalize_loc(fb_profile.get("city"), fb_profile.get("country"))
    s_loc = sim(cv_loc, fb_loc) if (cv_loc and fb_loc) else 0.15

    score = 0.75*s_name + 0.25*s_loc
    score = max(0.0, min(1.0, score * 0.90))
    if s_name < 0.55: flags.append("Name mismatch (Facebook)")
    if cv_loc and fb_loc and s_loc < 0.45: flags.append("Location mismatch (Facebook)")
    return float(score), flags

# 7) Verify one CV
async def verify_one_cv(cv_text: str, topk: int = 3) -> Dict[str, Any]:
    cv = await extract_cv_struct(cv_text)
    if not cv.get("name"):
        return {"score": 0.35, "best_profile": None, "flags": ["CV name not found"], "cv_extracted": cv}

    li_cands = await search_linkedin_candidates(cv, limit=8)

    li_ids = [c["id"] for c in li_cands if isinstance(c, dict) and isinstance(c.get("id"), int)]
    if li_ids:
        best_score, best, best_flags = -1.0, None, []
        for pid in li_ids[:topk]:
            prof = await get_linkedin_profile(pid)
            inter = await get_linkedin_interactions(pid)
            s, flags = score_linkedin(cv, prof, inter)
            if s > best_score:
                best_score = s
                best = {"source": "linkedin", "person_id": pid, "name": prof.get("name"), "headline": prof.get("headline")}
                best_flags = flags

        if best_score > 0.5 and len(best_flags) >= 4:
            best_score = max(0.0, best_score - 0.08)

        return {"score": float(best_score), "best_profile": best, "flags": best_flags[:8], "cv_extracted": cv}

    fb_cands = await search_facebook_candidates(cv, limit=8)

    fb_ids = [c["id"] for c in fb_cands if isinstance(c, dict) and isinstance(c.get("id"), int)]
    best_score, best, best_flags = -1.0, None, []
    for uid in fb_ids[:topk]:
        fb_prof = await get_facebook_profile(uid)
        s, flags = score_facebook(cv, fb_prof)
        if s > best_score:
            best_score = s
            best = {"source": "facebook", "user_id": uid, "display_name": fb_prof.get("display_name")}
            best_flags = flags

    if best_score < 0:
        return {"score": 0.25, "best_profile": None, "flags": ["No usable candidates on LinkedIn/Facebook"], "cv_extracted": cv}

    return {"score": float(best_score), "best_profile": best, "flags": best_flags[:8], "cv_extracted": cv}


# 8) Run all CVs -> scores
def _cv_num(file_name: str) -> int:
    m = re.findall(r"\d+", file_name or "")
    return int(m[-1]) if m else 10**9

all_cvs_sorted = sorted(all_cvs, key=lambda x: _cv_num(x.get("file", "")))

results = []
scores = []

for item in all_cvs_sorted:
    r = await verify_one_cv(item["text"], topk=3)
    results.append({"file": item["file"], **r})
    scores.append(float(r["score"]))

print("scores =", scores)

for r in results:
    print("=" * 90)
    print("FILE:", r["file"])
    print("SCORE:", r["score"])
    print("BEST_PROFILE:", r["best_profile"])
    print("FLAGS:", r["flags"])
    cvx = r.get("cv_extracted", {})
    print("EXTRACTED_NAME:", cvx.get("name"))
    print("EXTRACTED_LOCATION:", cvx.get("location"))
```

# Evaluation code

In the test phase, you will be given 5 CV files with fixed names:

    CV_1.pdf, CV_2.pdf, CV_3.pdf, CV_4.pdf, CV_5.pdf

Your system must process these CVs and output a list of 5 scores,
one score per CV, in the same order:

    scores = [s1, s2, s3, s4, s5]

Each score must be a float in the range [0, 1], representing the
reliability or confidence that the CV is valid (or meets the task criteria).

The ground-truth labels are binary:

    groundtruth = [0 or 1, ..., 0 or 1]

Each CV is evaluated independently using a threshold of 0.5:

- If score > 0.5 and groundtruth == 1 → Full credit
- If score ≤ 0.5 and groundtruth == 0 → Full credit
- Otherwise → No credit

In other words, 0.5 is the decision threshold.

- Each CV contributes equally.
- Final score = (number of correct decisions) / 5


```python
# =====================================================
#  Evaluation code
# =====================================================

def evaluate(scores, groundtruth, threshold=0.5):
    """
    scores: list of floats in [0, 1], length = 5
    groundtruth: list of ints (0 or 1), length = 5
    """
    assert len(scores) == 5
    assert len(groundtruth) == 5

    correct = 0
    decisions = []

    for s, gt in zip(scores, groundtruth):
        pred = 1 if s > threshold else 0
        decisions.append(pred)
        if pred == gt:
            correct += 1

    final_score = correct / len(scores)

    return {
        "decisions": decisions,
        "correct": correct,
        "total": len(scores),
        "final_score": final_score
    }
```

```python

groundtruth = [1, 1, 1, 0, 0] # Do not modify

result = evaluate(scores, groundtruth)
print(result)
```
